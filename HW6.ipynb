```python
#1
arr = np.array([[2,3,5],[4,6,8],[11,13,17],[7,10,13]]) 
def containsPrimes(arr):
    def is_prime(n):
        if n <= 1:
            return False
        if n <= 3:
            return True
        if n % 2 == 0 or n % 3 ==0:
            return False
        i = 5
        while i * i <= n:
            if n % i == 0 or n % (i + 2) == 0:
                return False
            i += 6 
        return True
    prime_rows = np.array([any(is_prime(n) for n in row) for row in arr])
    return prime_rows

print(containsPrimes(arr))
```

```python
#2.2
def alternating_rows(num_rows, num_cols):
    matrix = np.zeros((num_rows,num_cols))
    for i in range(num_rows):
        if (i + 1) % 2 != 0: 
            for j in range(num_cols):
                if (j + 1) % 2 != 0:
                    matrix[i,j]=1
    return matrix

print(alternating_rows(8,8))
```

```python
#3
universe = np.array(['galaxy', 'clusters'])
def expansion(text, num_spaces):
    arr = []
    for character in text:
        result = ''
        for letter in character:
            result += letter + ' ' * num_spaces
        arr.append(result)
    return arr

print(expansion(universe,2))
#done
```
